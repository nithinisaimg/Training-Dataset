```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import trim_mean, median_abs_deviation

df = pd.read_csv("state.csv")


print("First 5 Records")
print(df.head())


data = df[['Population', 'Murder.Rate']]

pearson_corr = data['Population'].corr(data['Murder.Rate'], method='pearson')

print("\nPearson Correlation Coefficient")
print("--------------------------------")
print(f"Pearson Correlation = {pearson_corr:.4f}")


corr_matrix = data.corr(method='pearson')

print("\nCorrelation Matrix")
print("------------------")
print(corr_matrix)

plt.figure(figsize=(6,5))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True
)

plt.title("Pearson Correlation Heatmap")
plt.show()


print("\nInterpretation")
print("----------------")

if pearson_corr > 0.7:
    print("Strong Positive Correlation")
elif pearson_corr > 0.3:
    print("Moderate Positive Correlation")
elif pearson_corr > 0:
    print("Weak Positive Correlation")
elif pearson_corr == 0:
    print("No Correlation")
elif pearson_corr > -0.3:
    print("Weak Negative Correlation")
elif pearson_corr > -0.7:
    print("Moderate Negative Correlation")
else:
    print("Strong Negative Correlation")


study_df = pd.DataFrame({
    'Hours Studied': [2, 4, 6, 8, 10, 12],
    'Marks': [45, 50, 65, 70, 85, 90]
})

pearson = study_df['Hours Studied'].corr(study_df['Marks'], method='pearson')
spearman = study_df['Hours Studied'].corr(study_df['Marks'], method='spearman')
kendall = study_df['Hours Studied'].corr(study_df['Marks'], method='kendall')

print("\nCorrelation Measures")
print("---------------------")
print("Pearson Correlation :", pearson)
print("Spearman's Rho      :", spearman)
print("Kendall's Tau       :", kendall)


summary = pd.DataFrame({
    'Mean': [df['Population'].mean()],
    'Trimmed Mean': [trim_mean(df['Population'], 0.10)],
    'Median': [df['Population'].median()],
    'Std Dev': [df['Population'].std()],
    'IQR': [df['Population'].quantile(0.75) - df['Population'].quantile(0.25)],
    'MAD': [median_abs_deviation(df['Population'])]
})

print("\nSummary Statistics")
print("-------------------")
print(summary)
```
