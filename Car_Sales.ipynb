```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Engine Size (L)': [1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8, 3.0, 3.5, 4.0],
    'Fuel Efficiency (MPG)': [42, 40, 38, 35, 32, 30, 27, 25, 22, 20],
    'Car Price': [12000, 15000, 18000, 22000, 25000, 28000, 32000, 35000, 42000, 50000]
}

df_cars = pd.DataFrame(data)

print("Car Dataset:")
print(df_cars)

correlation_matrix = df_cars.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)

plt.figure(figsize=(6, 5))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Matrix of Car Dataset")
plt.show()

sns.pairplot(df_cars, diag_kind="hist")
plt.show()

print("\nFinal Inference:")
print("1. Engine Size and Car Price have a strong positive correlation.")
print("   → Larger engine cars are generally more expensive.")

print("\n2. Engine Size and Fuel Efficiency have a strong negative correlation.")
print("   → As engine size increases, fuel efficiency decreases.")

print("\n3. Fuel Efficiency and Car Price have a strong negative correlation.")
print("   → More expensive cars in this dataset generally have lower fuel efficiency.")

print("\nConclusion:")
print("The strongest relationships are:")
print("- Engine Size ↔ Car Price : Strong Positive")
print("- Engine Size ↔ Fuel Efficiency : Strong Negative")
print("- Fuel Efficiency ↔ Car Price : Strong Negative")
print("These relationships indicate that larger engines increase vehicle cost but reduce fuel economy.")
```
